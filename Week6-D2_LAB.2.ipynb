```python
df_tweets = spark.read.format("avro").load("/Volumes/workspace/nf_schema/nf_data/tweets.avro")

df_tweets.printSchema()
df_tweets.show(5)
```

```python
from pyspark.sql.functions import col, count, upper

# 5 users with the highest number of tweets
df_top5 = df_tweets.groupBy("user.user_id").agg(count("tweet_id").alias("count")).orderBy("count", ascending=False).limit(5)
df_top5.display()

# Most commonly used words in all the tweets, excluding stopwords

# Filter the tweets that are in Spanish
df_ñ = df_tweets.filter(col("lang").rlike("sp|es|esp")).limit(6)
df_ñ.display()

# Which country has the highest number of tweets?
df_top1_twitcount = df_tweets.groupBy("user.country").agg(count("user.country").alias("twits")).orderBy("twits", ascending=False).limit(1)
df_top1_twitcount.display()

#Separate the tweets by a specific username ex: All tweets by Mark
print(dict(df_tweets.dtypes).get("user"))
df_username = df_tweets.filter(upper(col("user.username")).rlike("MARK"))
df_username.display()
df_display_name = df_tweets.filter(upper(col("user.display_name")).rlike("MARK"))
df_display_name.display()


```

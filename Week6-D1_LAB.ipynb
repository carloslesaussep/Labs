```python
df = spark.read.option("header", True).option("inferSchema", True).csv("/Volumes/workspace/nf_schema/nf_data/netflix_titles.csv")

#Get the schema
print("Schema:")
df.printSchema()

#Get the total entry amount registered in the dataset
print("Entry amount (Movies & Shows):")
df.count()

#Get the Nulls/missing values within director
print("Nulls/missing values within director:")
df.select("director").filter(df.director.isNull()).count()

#Get the ratings and the amount of movies and series registered in each rating
print("Ratings and the amount of movies and series registered in each rating:")
df.select("rating").groupBy("rating").count().orderBy("count", ascending=False).show()

#Get the top 10 listed_in genres by title count
print("Top 10 listed_in genres by title count:")
df.select("listed_in").groupBy("listed_in").count().orderBy("count", ascending=False).limit(10).show()

#Get the top 15 genres by title count
print("Top 15 genres by title count:")
df.select("listed_in").groupBy("listed_in").count().orderBy("count", ascending=False).limit(15).show()

#Get the type of available content 
print("Available content:")
df.select("type").groupBy("type").count().orderBy("count", ascending=False).show()

#Get the quantity of newly added contents by month and year (2026 only, month by month)
from pyspark.sql.functions import col, to_date, try_to_date

#df.select("date_added").to_date(col("date_added"), "MMMM d, yyyy").count().orderBy("count", ascending = False).show

#Get the average duration of TV Shows in season
print("Avg duration of TV shows:")
from pyspark.sql.functions import avg, col, regexp_extract, to_date
df.filter(col("type") == "TV Show").select(avg(regexp_extract(col("duration"), r"(\d+)", 1).try_cast("int")).alias("Average duration per show")).show()

#"Its Christmas! What movies/Tv shows are you binge watching this weekend?"
print("Christmas movie recommendation:")
actors_pattern = "Eddie Murphy|Jim Carrey|Arnold Schwarzenegger|Brendan Fraser"
df.select("title", "duration").filter(col("cast").rlike(actors_pattern)).filter(col("type") != "TV Show").show(truncate=False)

#Ryan Reinolds
print("Ryan Reynolds movies:")
df.select("title", "release_year").filter(col("cast").contains("Ryan Reynolds")).filter(col("type") != "TV Show").orderBy("release_year", ascending=False).show(truncate=False)

#Releases during COVID-19
#print("Releases during COVID-19:")
#df.select(try_to_date(col("date_added"), "MMMM d, yyyy").alias("date_added")).filter(year("date_added") == 2020).count()


#Movies of under two hours - Not working
#print("Movies of under two hours:")
#df.select("title", "duration").filter(col("type") == "Movie").filter(regexp_extract(col("duration"), r"(\d+)", 1).try_cast("int")).show

```
